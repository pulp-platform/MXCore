# Case 1: Sum >> Acc

```python
vector_size = 4
scale = 34
seed=709571

# Set the seed once
if seed is not None:
    random.seed(seed)

# Generate random FP9 vectors
random_fp9_a = [FP9.generate_random().value for _ in range(vector_size)]
random_fp9_b = [FP9.generate_random().value for _ in range(vector_size)]

random_fp9_a = [random_fp9_a[3], random_fp9_a[3], random_fp9_a[3], random_fp9_a[3]]
random_fp9_b = [random_fp9_b[3], random_fp9_b[3], random_fp9_b[3], random_fp9_b[3]]

# Generate a random FP32 value
random_fp32_s = FP32.generate_random()

# Overwrite the random FP32 value with a specific exponent value for testing
new_exponent = 14 + 127 # excess-127 notation
new_exponent_bin = format(new_exponent, '08b')
fp32_binary = '1' + new_exponent_bin + random_fp32_s.binary[9:]
random_fp32 = FP32.binary_to_float(fp32_binary)
random_fp32_s = FP32(random_fp32)

# Print generated vectors and FP32 value for reference
print(f"Random FP9 Vector A: {random_fp9_a}")
print(f"Random FP9 Vector B: {random_fp9_b}")
print(f"Random FP32 Value (s): {random_fp32_s.value}")

# Test the sum_fp9_sop_fp32 function with the generated vectors
sum_fp9_sop_fp32(random_fp9_a, random_fp9_b, random_fp32_s.value, scale)
```
